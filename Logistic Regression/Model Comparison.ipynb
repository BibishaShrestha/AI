#### 2. How does adding more features affect accuracy and recall?

The improvement depends on the quality and relevance of the added features.
Adding relevant features usually improves both accuracy and recall. With more information, the model can better distinguish between patients with and without heart disease. Accuracy increases because the model makes more correct predictions overall, while recall improves because the model becomes more capable of correctly identifying patients who actually have heart disease.

There is a trade-off between interpretability and performance when moving from a single-feature model to a multi-feature model. The Task 1 model is highly interpretable because predictions are based only on cholesterol levels, making it easy to understand the influence of the feature. And task 2 model is less straightforward to interpret because it considers many variables simultaneously. Despite the increased complexity, the model generally achieves better predictive performance and provides a more comprehensive assessment of heart disease risk. So, Task 1 offers simplicity and ease of interpretation, while task 2 provides higher accuracy and better overall prediction capability.
