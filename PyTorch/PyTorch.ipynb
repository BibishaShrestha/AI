## Tensor Creation Functions

PyTorch provides various functions to create tensors quickly for different purposes such as random initialization, zero initialization, and identity matrices.

1. torch.rand() - Creates a tensor with random values.
2. torch.zeros() - Creates a tensor filled with zeros.
3. torch.ones() - Creates a tensor filled with ones.
4. torch.eye() - Creates an identity matrix.
5. torch.arange() - Generates a sequence of values with a fixed step.
6. torch.linspace() - Generates evenly spaced values within a range.

```python
rand = torch.rand(3,4)
zeros = torch.zeros(3,4)
ones = torch.ones(3,4)
eye = torch.eye(3)
arange = torch.arange(start=0, end=10, step=2)
linspace = torch.linspace(start=0, end=10, steps=5)

print(f"rand: {rand}")
print(f"zeros: {zeros}")
print(f"ones: {ones}")
print(f"eye: {eye}")
print(f"arange: {arange}")
print(f"linspace: {linspace}")
```

## Tensor Arithmetic
Arithmetic operations can be performed directly on tensors. These operations are applied element-wise.

### `+` Operator

The `+` operator is used for element-wise addition between tensors.\
It provides a simple and readable syntax that is easy to understand.\
A new tensor is created to store the result of the addition.\
It automatically supports broadcasting when tensor shapes are compatible.\
It is the most commonly used method for tensor addition in PyTorch.

### `torch.add()`

`torch.add()` is a PyTorch function used for element-wise tensor addition.\
It produces the same result as the `+` operator in normal addition operations.\
It provides additional parameters such as `alpha` for scaling values before addition.\
It also supports the `out` parameter for storing results in a specified tensor.\
It is useful when greater control over memory usage and computations is required.


```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([3, 4, 5])

print(a+b)
print(a-b)
print(a*b)

# adding using add()
print(torch.add(a,b))

print(torch.add(a, b, alpha=2))
```

## @ vs. * for Tensor Multiplication

#### * (Element-wise Multiplication)
The * operator performs element-wise multiplication between tensors.\
Each element of the first tensor is multiplied by the corresponding element of the second tensor.\
The tensors must have the same shape or compatible shapes through broadcasting.\
The output tensor has the same shape as the input tensors.\
It is commonly used for scaling values and feature-wise operations.
#### @ (Matrix Multiplication)
The @ operator performs matrix multiplication between tensors.\
It follows the rules of linear algebra, where inner dimensions must match.\
For matrices, the number of columns in the first tensor must equal the number of rows in the second tensor.\
The resulting tensor shape is determined by the outer dimensions.\
It is widely used in deep learning for operations such as neural network computations and linear transformations.

```python
# Element-wise multiplication (*)
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])

print(f"Tensor A: {a}")
print(f"Tensor B: {b}")
print(f"Element-wise multiplication (A * B): {a * b}")

# Matrix multiplication (@)
mat_a = torch.tensor([[1, 2],
                      [3, 4]])
mat_b = torch.tensor([[5, 6],
                      [7, 8]])

print(f"\nMatrix A: {mat_a}")
print(f"Matrix B: {mat_b}")
print(f"Matrix multiplication (A @ B): {mat_a @ mat_b}")

# Example with 1D tensors (dot product)
vec_a = torch.tensor([1, 2, 3])
vec_b = torch.tensor([4, 5, 6])

print(f"\nVector A: {vec_a}")
print(f"Vector B: {vec_b}")
print(f"Dot product (Vector A @ Vector B): {vec_a @ vec_b}")
```

## Memory Behavior

The main difference is how they handle memory sharing.

A view means the new tensor uses the same underlying memory as the original tensor.
A copy means a new separate memory block is created.

### tensor.view()
Always tries to return a view (no new memory allocation).
Both tensors share the same data in memory.
So, changing one tensor affects the other.
Requires the tensor to be contiguous in memory (data stored in a clean, continuous layout).
If not contiguous (e.g., after transpose), view() will fail unless you call .contiguous() first, which may create a copy.

Key idea: Strict but memory-efficient

### tensor.reshape()
More flexible than view().
First tries to return a view if possible.
If the memory layout is not suitable, it automatically creates a copy.
Works even with non-contiguous tensors.

Key idea: Flexible but may use extra memory

```python
t = torch.arange(11, 20, 1)  # values from 10 to 18

print(t.reshape(3, 3))
print(t.view(3, 3))
print(t)
```

#### squeeze() and unsqueeze()
squeeze() removes dimensions of size 1, while unsqueeze() adds a dimension of size 1 at a specified position.

```python
x = torch.tensor([[[4.0], [5.0], [6.0]]])

print(x.shape)
print(x.squeeze())        
print(x.squeeze().shape)

print(x.unsqueeze(1)) 
```

#### torch.stack()
torch.stack() combines multiple tensors into a single tensor by creating a new dimension.

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
c = torch.tensor([7, 8, 9])

print(torch.stack([a, b, c], dim=0)) 
print(torch.stack([a, b, c], dim=1)) 
```

## Indexing and Aggregation
Indexing allows access to specific elements of a tensor, while aggregation functions such as min(), max(), mean(), and sum() provide summary statistics.

```python
t = torch.tensor([[2., 4., 7.],
                  [1., 9., 8.]])

print(t.min(), t.max())
print(t.mean(), t.sum())
print(t.argmin(), t.argmax())

print(t[1])       
print(t[:, 0])     
print(t[0, 2]) 
```

## Datatype and dtype Casting
Every tensor has a data type. Type casting is used to convert tensors from one data type to another when required.

```python
x = torch.tensor([1, 2, 3])
print(x.dtype)   # torch.int64 by default

f = x.type(torch.float32)
print(f)
print(f.dtype)

# or specify at creation
y = torch.tensor([1.0, 2.0], dtype=torch.float16)
print(y.dtype)
y = torch.tensor([2.6, 4.7, 2.5], dtype=torch.float64)
print(y)
print(y.dtype)
```

## Moving to GPU
Why GPUs over CPUs for Machine Learning?

#### CPU (Central Processing Unit)
Designed for general-purpose computing\
Has few but very powerful cores\
Optimized for sequential processing (one task after another)\
Handles complex logic and branching well\
Good at low-latency tasks (fast single operations)


#### GPU (Graphics Processing Unit)
Designed for massively parallel processing\
Has thousands of smaller cores\
Excels at performing many identical operations simultaneously\
Best for mathematical and matrix-heavy computations\
Optimized for high-throughput tasks

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

t = torch.tensor([1., 2., 3.])
t_gpu = t.to(device)
print(t_gpu.device)

# move back to CPU (needed before NumPy conversion)
t_cpu = t_gpu.cpu()
print(t_cpu.numpy())
```
